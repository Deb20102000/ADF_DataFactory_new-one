```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.functions import explode, col
spark = SparkSession.builder.appName("MetadataPipeline").getOrCreate()
```

```python
dbutils.widgets.text("entity_name", "")

entity_name = dbutils.widgets.get("entity_name")

print(f"Entity received from ADF: {entity_name}")
```

```python
landing_path = f"abfss://landing@str2052.dfs.core.windows.net/{entity_name.lower()}/"

print(landing_path)
```

```python
df = spark.read.format("json").load(landing_path)

display(df)
```

```python
df.printSchema()
```

```python
print(f"Entity: {entity_name}")
print(df.columns)

df1 = df.select(
    explode(col(entity_name.lower())).alias("record")
)
```

```python
from pyspark.sql.functions import explode

array_column = entity_name.lower()

df1 = df.select(
    explode(array_column).alias("record")
)
```

```python
bronze_df = df1.select("record.*")
```

```python
from pyspark.sql.functions import current_timestamp, lit

bronze_df = bronze_df \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_system", lit("DummyJSON")) \
    .withColumn("entity_name", lit(entity_name))
```

```python
display(bronze_df)
```

```python
bronze_path = f"abfss://bronze@str2052.dfs.core.windows.net/{entity_name.lower()}/"

print(bronze_path)
```

```python
bronze_df.write \
    .format("delta") \
    .mode("overwrite") \
    .save(bronze_path)
```

```python
display(dbutils.fs.ls(bronze_path))
```

```python
bronze_read_df = spark.read.format("delta").load(bronze_path)

display(bronze_read_df)
```

```python
print(bronze_read_df.count())
```
